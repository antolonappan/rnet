## Read an existing simulation suite


```python
from pathlib import Path
from rnet.cnn.config import DataConfig
from rnet.cnn.data import SimulationDataset

base = Path('/path/to/your/simulation/base')
manifest = base / 'unified_sat_parallel_v1' / 'recipes' / 'uniform_zero.json'
if manifest.is_file():
    data_config = DataConfig(manifest=manifest, frequencies=(93, 145), variant='nofg')
    dataset = SimulationDataset(data_config)
    print(dataset.summary())
else:
    print('Set base to a simulation suite before loading maps.')
```

## Create a simulator


```python
from rnet.sims import UnifiedSATSimulator

inputs = [base / 'data' / name for name in ('camb.ini', 'hitsmap.fits.gz', 'mask.fits', 'mask_apo.fits', 'mask_padded.fits')]
if all(path.is_file() for path in inputs):
    simulator = UnifiedSATSimulator(base, nside_cmb=1024, nside_out=128, lmax_out=300)
    print('Simulator ready:', simulator.root)
else:
    print('Add simulation inputs under', base / 'data', 'to construct a simulator.')
```
